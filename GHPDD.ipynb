# After Inspection and Cleaning of Data 

- I found out that there is no missing values in the dataset.
- The dataset was collected from different countries around the world.
- There are three furnishing_status in this dataset, which are Semi-Furnished, Fully-Furnished and Unfurnished.
- I found out that in the dataset there are more Fully-Furnished Houses.

# Analysis by Furnishing_status

```python
furnishing_status_counts = df['furnishing_status'].value_counts()
print(furnishing_status_counts)
```

```python
plt.figure(figsize=(8,8))
plt.pie(furnishing_status_counts, labels=furnishing_status_counts.index, autopct='%1.1f%%')
plt.title('Distribution of Furnishing Status')
plt.show()
```

# Analysis by Price

```python
country_price = df.groupby('country')['price'].mean().reset_index()

top_10_countries = country_price.nlargest(10, 'price')

plt.figure(figsize=(12,8))
sns.barplot(x='country', y='price', data=top_10_countries)
plt.title('Top 10 Countries by Price')
plt.xlabel('Country')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.show()
```

# Analysis by Loan_amount

```python
country_loan_amount = df.groupby('country')['loan_amount'].sum().reset_index()

top_10_countries = country_loan_amount.nlargest(10, 'loan_amount')

plt.figure(figsize=(12,8))
sns.barplot(x='country', y='loan_amount', data=top_10_countries)
plt.title('Top 10 Countries by Loan Amount')
plt.xlabel('Country')
plt.ylabel('Loan Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
max_loan_amount = df['loan_amount'].max()
print(f"The maximum loan amount is: {max_loan_amount}")
```

```python
max_loan_row = df.loc[df['loan_amount'].idxmax()]
print(max_loan_row)
```

# Analysis by Satisfaction_score

```python
country_satisfaction = df.groupby('country')['satisfaction_score'].mean().reset_index()
```

```python
top_countries = country_satisfaction.nlargest(10, 'satisfaction_score')
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='country', y='satisfaction_score', data=top_countries)
plt.title('Top 10 Countries by Satisfaction Score')
plt.xlabel('Country')
plt.ylabel('Satisfaction Score')
plt.xticks(rotation=45)
plt.show()
```

```python
top_countries_list = top_countries['country'].tolist()
df_top_countries = df[df['country'].isin(top_countries_list)]

plt.figure(figsize=(12,8))
sns.boxplot(x='country', y='satisfaction_score', data=df_top_countries)
plt.title('Distribution of Satisfaction Scores for Top 10 Countries')
plt.xlabel('Country')
plt.ylabel('Satisfaction Score')
plt.xticks(rotation=45)
plt.show()
```

# Analysis by Neighbourhood_rating

```python
neighbourhood_rating_counts = df['neighbourhood_rating'].value_counts().head(10)
print(neighbourhood_rating_counts)
```

```python
top_countries = df['country'].value_counts().head(10)
top_neighbourhood_ratings = df['neighbourhood_rating'].value_counts().head(10)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_countries.index, y=top_countries.values)
plt.title('Top 10 Countries')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_neighbourhood_ratings.index, y=top_neighbourhood_ratings.values)
plt.title('Top 10 Neighborhood Ratings')
plt.xlabel('Neighborhood Rating')
plt.ylabel('Count')
plt.show()
```

```python
country_neighbourhood_rating = df.groupby(['country', 'neighbourhood_rating']).size().unstack().fillna(0)
country_neighbourhood_rating = country_neighbourhood_rating.loc[country_neighbourhood_rating.sum(axis=1).sort_values(ascending=False).head(10).index]

plt.figure(figsize=(12,8))
sns.heatmap(country_neighbourhood_rating, annot=True, cmap='Blues')
plt.title('Relationship Between Country and Neighborhood Rating')
plt.xlabel('Neighborhood Rating')
plt.ylabel('Country')
plt.show()
```

# Analysis by Countries with Previous Owners

```python
# Count the number of previous owners for each country
country_previous_owners = df['country'].value_counts().reset_index()
country_previous_owners.columns = ['country', 'count']

# Get the top 10 countries with the most previous owners
top_10_countries = country_previous_owners.nlargest(10, 'count')

# Plot the top 10 countries
plt.figure(figsize=(12,8))
sns.barplot(x='country', y='count', data=top_10_countries)
plt.title('Top 10 Countries by Number of Previous Owners')
plt.xlabel('Country')
plt.ylabel('Number of Previous Owners')
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.show()
```

```python
crosstab = pd.crosstab(df['country'], df['previous_owners'])

plt.figure(figsize=(12,8))
sns.heatmap(crosstab, annot=True, cmap='Blues')
plt.title('Relationship between Countries and Previous Owners')
plt.xlabel('Previous Owner')
plt.ylabel('Country')
plt.show()
```
